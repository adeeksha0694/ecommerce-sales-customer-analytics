```python
# Imports

import pandas as pd
```

```python
# Load Data

df = pd.read_excel(r"C:\Users\Adeeksha\OneDrive\Documents\Data Analyst\Project\E-Commerce Sales & Customer Analytics\Data\sales_raw_500.xlsx")

df
```

```python
# removing duplicates

df = df.drop_duplicates()
df
```

```python
# fix blank columns

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df
```

```python
# dropping na columns

df = df.dropna(subset = ['Order_Date'])
df
```

```python
# creating profit date and month columns 

df['Profit'] = df['Sales'] - df['Cost']
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month

df
```

```python
# saving the data in csv format

df.to_csv('Sales_Cleaned.csv', index=False)
```
